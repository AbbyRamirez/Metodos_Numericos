```python
import numpy as np

def metodo_LU(K, F, n):
    # Crear L y U
    L = []
    U = []
    n_val = n
    for i in range(n_val):
        fila_L = []
        fila_U = []
        for j in range(n_val):
            fila_U.append(0.0)
            if (i == j):
                fila_L.append(1.0)
            else:
                fila_L.append(0.0)
        L.append(fila_L)
        U.append(fila_U)

    # Factorizacion LU
    for i in range(n_val):
        for k in range(i, n_val):
            suma = 0
            for j in range(i):
                suma += L[i][j] * U[j][k]
            U[i][k] = K[i][k] - suma

        for k in range(i+1, n_val):
            suma = 0
            for j in range(i):
                suma += L[k][j] * U[j][i]
            L[k][i] = (K[k][i] - suma) / U[i][i]

    # CALCULAR DETERMINANTE
    determinante = 1.0
    for i in range(n_val):
        determinante *= U[i][i]

    # Ly = F
    y = [0.0]*n_val
    for i in range(n_val):
        suma = 0
        for j in range(i):
            suma += L[i][j]*y[j]
        y[i] = F[i] - suma

    # Ux = y
    x = [0.0]*n_val
    for i in range(n_val-1, -1, -1):
        suma = 0
        for j in range(i+1, n_val):
            suma += U[i][j]*x[j]
        x[i] = (y[i]-suma)/U[i][i]

    # det
    print(f"Determinante: {determinante}")

    return x, determinante

def crear_matrices_grandes(n):

    # Generar una matriz
    K_large = (np.random.rand(n, n) * 10).tolist()
    # Generar un vector F
    F_large = (np.random.rand(n) * 10).tolist()
    return K_large, F_large

#  tamaño de la matriz y el vector
n_grande_personalizado = 100

# Crear las matrices
K_generada, F_generado = crear_matrices_grandes(n_grande_personalizado)


# Llamar a la función metodo_LU
x_sol_generada, det_val_generado = metodo_LU(K_generada, F_generado, n_grande_personalizado)

#Script inicial por Rachel Morera (mi verdadero real :D), se uso gemini en colab para debuggear (era un un frankenstein y un desorden)
```
